### Task 1: Object Oriented Grocery Manager

```python
class GroceryManager:
  def __init__(self):
    self.items = []

  def add_item(self, item, quantity, price):
    new_item = {'item': item, 'quantity': quantity, 'price': price}
    self.items.append(new_item)
    print('item added')

  def remove_item(self, item):
    for i in self.items:
      if i['item'] == item:
        self.items.remove(i)
        print(f'{item} removed')
        return
    print(f'{item} not found')



  def view_list(self):
    for i in self.items:
      print(i)

  def calculate_total(self):
    total = 0
    for i in self.items:
      total+= i['quantity'] *i['price']
    return f'Total: {total}'
```

```python
grocery = GroceryManager()

grocery.add_item('banana', 12, 1.2)
grocery.add_item('apple', 20, 2.2)
grocery.add_item('mango', 10, 2.8)

grocery.view_list()
```

```python
grocery.remove_item('apple')
grocery.view_list()
```

```python
grocery.calculate_total()
```

### Task 2: Advanced Student Record System

```python
students = {
    73: {'Name': 'Rayyan', 'Major': 'Artificial Intelligence', 'Grades': [88, 86, 92, 91]},
    74: {'Name': 'Malik', 'Major': 'Artificial Intelligence', 'Grades': [90, 87, 94, 89]},
    75: {'Name': 'Arham', 'Major': 'Computer Science', 'Grades': [82, 85, 88, 90]},
    76: {'Name': 'Hamza', 'Major': 'Computer Science', 'Grades': [91, 89, 86, 93]},
    77: {'Name': 'Bobby', 'Major': 'Artificial Intelligence', 'Grades': [85, 90, 88, 92]}
}
```

```python
def highest_avg_grade(students):
  highest_avg_grade_ID = 0
  highest_avg_grade_value = 0
  for key, value in students.items():
    avg_grade = sum(value['Grades'])/len(value['Grades'])
    if avg_grade > highest_avg_grade_value:
      highest_avg_grade_value = avg_grade
      highest_avg_grade_ID = key
  print(f'{students[highest_avg_grade_ID]['Name']}: {highest_avg_grade_value}')


def search_major(students, major):
  for key, value in students.items():
    if value['Major'] == major:
      print(value['Name'])
```

```python
highest_avg_grade(students)
```

```python
search_major(students, 'Artificial Intelligence')
```

### Task 5: Advanced Blurring & NumPy ROI Extraction

```python
image_path = "image.jpg"
image = cv2.imread(image_path)

if image is None:
    print("image not found")
else:
    image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
```

```python
blurred = cv2.GaussianBlur(image_rgb, (25, 25), 0)
```

```python
height, width = image_rgb.shape[:2]

center_x = width // 2
center_y = height // 2

print("Image dimensions:", width, "x", height)
print("Center coordinates:", (center_x, center_y))
```

```python
half = 150

x_start = center_x - half
x_end = center_x + half
y_start = center_y - half
y_end = center_y + half
```

```python
roi_original = image_rgb[y_start:y_end, x_start:x_end]

roi_blurred = blurred[y_start:y_end, x_start:x_end]
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 6))

axes[0].imshow(roi_original)
axes[0].axis("off")
axes[0].set_title("ROI", fontsize=14)

axes[1].imshow(roi_blurred)
axes[1].axis("off")
axes[1].set_title("Blurred", fontsize=14)

plt.show()
```

### Task 7: Matrix-Based Rotation & Adaptive Thresholding

```python
image = cv2.imread("document.png", cv2.IMREAD_GRAYSCALE)
```

```python
_, global_thresh = cv2.threshold(image,127,255,cv2.THRESH_BINARY)
```

```python
adaptive_thresh = cv2.adaptiveThreshold(image,255,cv2.ADAPTIVE_THRESH_GAUSSIAN_C,cv2.THRESH_BINARY,11,2)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 6))

axes[0].imshow(global_thresh, cmap="gray")
axes[0].axis("off")
axes[0].set_title("Global Binary Threshold")

axes[1].imshow(adaptive_thresh, cmap="gray")
axes[1].axis("off")
axes[1].set_title("Adaptive Threshold")
```

```python
height, width = adaptive_thresh.shape
center = (width // 2, height // 2)
```

```python
rotation_matrix = cv2.getRotationMatrix2D(center,45,0.8)
rotated = cv2.warpAffine(adaptive_thresh,rotation_matrix,(width,height))
```

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 5))

axes[0].imshow(global_thresh, cmap="gray")
axes[0].axis("off")
axes[0].set_title("Global Binary Threshold")

axes[1].imshow(adaptive_thresh, cmap="gray")
axes[1].axis("off")
axes[1].set_title("Adaptive Threshold")

axes[2].imshow(rotated, cmap="gray")
axes[2].axis("off")
axes[2].set_title("Adaptive Threshold Rotated")
```

### Task 8: Masking with Bitwise Logic

```python
image1 = cv2.imread("Sukuna.jpg")
image2 = cv2.imread("image.jpg")
```

```python
height, width = image1.shape[:2]
image2 = cv2.resize(image2, (width, height))
```

```python
mask = np.zeros((height, width), dtype=np.uint8)
```

```python
center = (width // 2, height // 2)

radius = min(width, height) // 4

cv2.circle(mask,center,radius,255,-1)
```

```python
foreground = cv2.bitwise_and(image1,image1,mask=mask)
```

```python
inverse_mask = cv2.bitwise_not(mask)
```

```python
background = cv2.bitwise_and(image2,image2,mask=inverse_mask)
```

```python
final_image = cv2.bitwise_or(foreground,background)
```

```python
final_rgb = cv2.cvtColor(final_image,cv2.COLOR_BGR2RGB)

image1_rgb = cv2.cvtColor(image1, cv2.COLOR_BGR2RGB)
image2_rgb = cv2.cvtColor(image2, cv2.COLOR_BGR2RGB)
mask_rgb = cv2.cvtColor(mask, cv2.COLOR_GRAY2RGB)
foreground_rgb = cv2.cvtColor(foreground, cv2.COLOR_BGR2RGB)
background_rgb = cv2.cvtColor(background, cv2.COLOR_BGR2RGB)
final_rgb = cv2.cvtColor(final_image, cv2.COLOR_BGR2RGB)

fig, axes = plt.subplots(2,3,figsize=(15, 10))

axes[0, 0].imshow(image1_rgb)
axes[0, 0].set_title("Image 1")
axes[0, 0].axis("off")

axes[0, 1].imshow(image2_rgb)
axes[0, 1].set_title("Image 2")
axes[0, 1].axis("off")

axes[0, 2].imshow(mask, cmap="gray")
axes[0, 2].set_title("Binary Mask")
axes[0, 2].axis("off")

axes[1, 0].imshow(foreground_rgb)
axes[1, 0].set_title("Image 1 - Circle")
axes[1, 0].axis("off")

axes[1, 1].imshow(background_rgb)
axes[1, 1].set_title("Image 2 - Background")
axes[1, 1].axis("off")

axes[1, 2].imshow(final_rgb)
axes[1, 2].set_title("Final Combined Image")
axes[1, 2].axis("off")

plt.tight_layout()
plt.show()
```
